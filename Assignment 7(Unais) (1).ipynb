# Question 1: Employee Data Bank (Using OOP)


```python
class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print(f"Name               : {self.name}")
        print(f"Employee Number    : {self.number}")
        print(f"Address            : {self.address}")
        print(f"Salary             : {self.salary}")
        print(f"Post               : {self.post}")
        print(f"Year of Joining    : {self.year_of_joining}")
        print(f"Education          : {self.education}")
        print(f"Years of Experience: {self.year_of_experience}")
        print("-" * 45)

    def update_year_of_experience(self, years=1):
        self.year_of_experience += years
        print(f"{self.name}'s experience updated to {self.year_of_experience} years.")

    def increase_salary(self, amount):
        self.salary += amount
        print(f"{self.name}'s salary increased by {amount}. New salary = {self.salary}")


employee101 = Employee("Alice", 101, "123 Main Street", 50000,
                       "Software Engineer", 2020,
                       "BSc Computer Science", 3)

print("=== Four individual attributes of employee101 ===")
print("Name   :", employee101.name)
print("Number :", employee101.number)
print("Address:", employee101.address)
print("Salary :", employee101.salary)

print("\n=== Calling all methods on employee101 ===")
employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience()
employee101.describe_employee()

employees = [
    Employee("Bob",     102, "45 Oak Avenue",   60000, "Data Analyst",        2018, "MSc Data Science",          5),
    Employee("Charlie", 103, "78 Pine Road",    55000, "HR Manager",          2019, "MBA",                       4),
    Employee("Diana",   104, "12 Cedar Lane",   70000, "Project Manager",     2016, "PMP",                       7),
    Employee("Ethan",   105, "90 Maple Drive",  48000, "Accountant",          2021, "BCom",                      2),
    Employee("Fiona",   106, "33 Birch Street", 52000, "QA Engineer",         2020, "BSc IT",                    3),
    Employee("George",  107, "67 Elm Court",    65000, "DevOps Engineer",     2017, "BSc Computer Science",      6),
    Employee("Hannah",  108, "21 Willow Way",   47000, "Support Engineer",    2022, "BSc IT",                    1),
    Employee("Ian",     109, "5 Aspen Place",   72000, "Senior Developer",    2015, "MSc Software Engineering",  8),
    Employee("Julia",   110, "88 Chestnut Blvd",51000, "Marketing Specialist",2021, "MBA",                       2),
    Employee("Kevin",   111, "19 Spruce Street",58000, "Network Engineer",    2019, "BSc Network Security",      4),
]

print("\n=== describe_employee() for all 10 employees ===")
for emp in employees:
    emp.describe_employee()

print("=== Updating employees ===")
employees[0].increase_salary(3000)
employees[0].update_year_of_experience(1)

employees[1].increase_salary(2500)
employees[1].update_year_of_experience(2)

employees[2].increase_salary(4000)
employees[2].update_year_of_experience()

print("\n=== Updated details of first 3 employees ===")
for emp in employees[:3]:
    emp.describe_employee()
```
